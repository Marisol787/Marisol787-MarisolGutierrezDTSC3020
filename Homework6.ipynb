```python
#add the libraries first
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Suppress future warnings from scikit-learn
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
# Define the 2D dataset
data = np.array([
    [2, 10], # A
    [2, 5],  # B
    [8, 4],  # C
    [5, 8],  # D
    [7, 5],  # E
    [6, 4]   # F
])

# Define point labels
point_labels = ['A', 'B', 'C', 'D', 'E', 'F']

print("Dataset points:\n", data)
```

```python
# Apply K-Means clustering with K = 2
kmeans = KMeans(n_clusters=2, random_state=0, n_init=10) # n_init to avoid warning
kmeans.fit(data)

clusters = kmeans.labels_

centroids = kmeans.cluster_centers_

print("Final Centroid Coordinates:")
for i, centroid in enumerate(centroids):
    print(f"Centroid {i+1}: {centroid}")
```

```python

plt.figure(figsize=(8, 6))

colors = ['red', 'blue']

for i in range(len(data)):
    plt.scatter(data[i, 0], data[i, 1], color=colors[clusters[i]], s=100)
    plt.text(data[i, 0] + 0.3, data[i, 1] + 0.3, point_labels[i], fontsize=12)

plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=200, color='black', label='Centroids')

plt.title('K-Means Clustering with K=2')
plt.xlabel('X-coordinate')
plt.ylabel('Y-coordinate')
plt.grid(True)
plt.legend()
plt.show()
```
